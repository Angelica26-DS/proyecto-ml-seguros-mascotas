## 5.3 Variable objetivo y riesgo de Data Leakage

### Objetivo

Verificar la relación existente entre la variable objetivo (`precio_mensual`) y la variable `precio_anual`, con el fin de identificar posibles riesgos de **Data Leakage** antes del entrenamiento del modelo.

### Implementación

```python
comparacion_precios = pd.DataFrame({
    "precio_mensual": df["precio_mensual"],
    "precio_mensual_x12": (df["precio_mensual"] * 12).round(2),
    "precio_anual": df["precio_anual"]
})

comparacion_precios.head(10)
```

```python
coinciden = (
    (df["precio_mensual"] * 12)
    .round(2)
    .equals(df["precio_anual"].round(2))
)

resultado_leakage = pd.DataFrame({
    "Comprobación": ["¿precio_anual = precio_mensual × 12?"],
    "Resultado": [coinciden]
})

resultado_leakage
```

### Análisis complementario

La comprobación inicial mostró que **precio_anual** no corresponde exactamente al valor de **precio_mensual × 12**.

Por este motivo, se realiza un análisis adicional para comprender la relación existente entre ambas variables y determinar si **precio_anual** sigue representando una transformación derivada de la variable objetivo.

```python
comparacion_precios["descuento"] = (
    comparacion_precios["precio_mensual_x12"]
    - comparacion_precios["precio_anual"]
).round(2)

comparacion_precios["descuento"].describe()
```

```python
relacion = (
    df["precio_anual"] / df["precio_mensual"]
).round(4)

relacion.describe()
```

### Conclusiones

La comprobación realizada muestra que **precio_anual** no corresponde exactamente al valor de **precio_mensual × 12**.

Sin embargo, el análisis evidencia que ambas variables mantienen una relación prácticamente constante, con un factor medio cercano a **11,42**, lo que indica que **precio_anual** se obtiene mediante una transformación comercial derivada directamente del **precio_mensual**.

La reducida variabilidad observada en dicha relación sugiere la existencia de una regla de negocio estable, posiblemente asociada a descuentos o condiciones comerciales aplicadas a las cotizaciones anuales.

### Decisión

La variable **precio_anual** será eliminada del conjunto de variables predictoras.

Aunque no representa exactamente el valor de **precio_mensual × 12**, continúa siendo una variable derivada directamente de la variable objetivo y su utilización durante el entrenamiento introduciría **Data Leakage**, proporcionando al modelo información que no estaría disponible en una nueva cotización.

>*Hallazgo*
>
> El análisis permitió identificar que **precio_anual** no es simplemente el resultado de multiplicar el precio mensual x 12, sino que incorpora una transformación comercial consistente. Aun así, mantiene una dependencia directa con la variable objetivo, por lo que no debe utilizarse como variable predictora.

### Objetivo

### Implementación

### Función: `revisar_variable_categorica()`

Con el fin de evitar la repetición de código durante el análisis de las variables categóricas, se define una función reutilizable que resume automáticamente la información más relevante de cada variable.

La función muestra:

- El nombre de la variable.
- El número de valores únicos.
- El número de valores nulos.
- La frecuencia absoluta de cada categoría.
- El porcentaje que representa cada categoría respecto al total de registros.
Esta función será utilizada durante la revisión de todas las variables categóricas del proyecto.

```python
def revisar_variable_categorica(df, columna):
    """
    Analiza una variable categórica mostrando un resumen estadístico
    y la distribución de sus categorías.

    Esta función se utiliza durante la fase de revisión del dataset
    para apoyar la toma de decisiones del preprocesamiento.

    Parámetros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene la información.

    columna : str
        Nombre de la variable categórica que se desea analizar.

    Salida
    -------
    Muestra dos tablas:

    1. Resumen general
        - Nombre de la variable.
        - Número de valores únicos.
        - Número de valores nulos.

    2. Distribución de categorías
        - Frecuencia absoluta.
        - Porcentaje respecto al total de registros.
    """

    # ==========================================================
    # Resumen general de la variable
    # ==========================================================
    resumen = pd.DataFrame({
        "Variable": [columna],
        "Valores únicos": [df[columna].nunique(dropna=False)],
        "Valores nulos": [df[columna].isna().sum()]
    })

    display(resumen)

    # ==========================================================
    # Distribución de frecuencias y porcentaje
    # ==========================================================
    frecuencias = (
        df[columna]
        .value_counts(dropna=False)
        .rename("Frecuencia")
        .to_frame()
    )

    frecuencias["Porcentaje (%)"] = (
        frecuencias["Frecuencia"] / len(df) * 100
    ).round(2)

    display(frecuencias)
```

### Utilidad de la función

La función `revisar_variable_categorica()` permite automatizar el análisis de las variables categóricas del conjunto de datos, facilitando la identificación de:

- categorías poco frecuentes;
- valores nulos;
- distribuciones desequilibradas;
- posibles situaciones que puedan requerir tratamiento durante el preprocesamiento.

El uso de una función reutilizable garantiza un análisis homogéneo para todas las variables y mejora la legibilidad, mantenibilidad y escalabilidad del notebook.

```python
for variable in variables_categoricas:
    revisar_variable_categorica(df, variable)
```

Las variables `grupo_raza_nombre` y `pet_race_name` no presentan valores nulos y muestran una distribución controlada, coherente con la naturaleza parcialmente simulada del conjunto de datos.

### Función auxiliar: `revisar_variable_binaria()`

Con el objetivo de estandarizar el análisis de las variables binarias, se define una función reutilizable que resume sus principales características y facilita la revisión del dataset durante la fase de preprocesamiento.

La función muestra:

- Información general de la variable.
- Número de categorías.
- Valores nulos.
- Distribución de frecuencias.
- Porcentaje de cada categoría.
- Validación de que la variable contiene únicamente dos categorías no nulas.

```python
def revisar_variable_binaria(df, columna):
    """
    Analiza una variable binaria mostrando un resumen de sus categorías,
    valores nulos y distribución.

    Parámetros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene la información.

    columna : str
        Nombre de la variable binaria que se desea analizar.

    Salida
    -------
    Muestra dos tablas:

    1. Resumen general
        - Nombre de la variable.
        - Número de valores únicos no nulos.
        - Número de valores nulos.
        - Validación de estructura binaria.

    2. Distribución de categorías
        - Frecuencia absoluta.
        - Porcentaje respecto al total de registros.
    """

    # Resumen general de la variable
    valores_unicos = df[columna].dropna().nunique()

    resumen = pd.DataFrame({
        "Variable": [columna],
        "Valores únicos no nulos": [valores_unicos],
        "Valores nulos": [df[columna].isna().sum()],
        "¿Es binaria?": [valores_unicos == 2]
    })

    display(resumen)

    # Distribución de frecuencias y porcentaje
    frecuencias = (
        df[columna]
        .value_counts(dropna=False)
        .rename("Frecuencia")
        .to_frame()
    )

    frecuencias["Porcentaje (%)"] = (
        frecuencias["Frecuencia"] / len(df) * 100
    ).round(2)

    display(frecuencias)
```

```python
for variable in variables_binarias:
    revisar_variable_binaria(df, variable)
```

Las cuatro variables binarias presentan una estructura consistente, sin valores nulos y con únicamente dos categorías válidas, por lo que son adecuadas para ser utilizadas en el proceso de modelado.

Se observa que:

- `pet_sterilized` mantiene la misma distribución, con un 60 % de mascotas esterilizadas y un 40 % no esterilizadas.

No se detectan inconsistencias ni categorías inesperadas en ninguna de las variables analizadas.

Durante el preprocesamiento:

```python
def revisar_variable_ordinal(df, columna):
    """
    Analiza una variable ordinal mostrando un resumen estadístico
    y la distribución de sus categorías.

    Esta función se utiliza durante la fase de revisión del dataset
    para apoyar la toma de decisiones del preprocesamiento.

    Parámetros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene la información.

    columna : str
        Nombre de la variable ordinal.

    Salida
    -------
    Muestra dos tablas:

    1. Resumen general.
    2. Distribución de categorías con frecuencia y porcentaje.
    """

    # ==========================================================
    # Resumen general
    # ==========================================================

    resumen = pd.DataFrame({
        "Variable": [columna],
        "Valores únicos": [df[columna].nunique(dropna=False)],
        "Valores nulos": [df[columna].isna().sum()]
    })

    display(resumen)

    # ==========================================================
    # Distribución
    # ==========================================================

    frecuencias = (
        df[columna]
        .value_counts(dropna=False)
        .rename("Frecuencia")
        .to_frame()
    )

    frecuencias["Porcentaje (%)"] = (
        frecuencias["Frecuencia"] / len(df) * 100
    ).round(2)

    display(frecuencias)
```

### Función auxiliar: `revisar_variable_temporal()`

Se define una función reutilizable para analizar variables temporales.

La función resume:

- Tipo de dato.
- Número de fechas distintas.
- Fecha mínima.
- Fecha máxima.

```python
def revisar_variable_temporal(df, columna):
    """
    Analiza una variable temporal mostrando un resumen de su estructura.

    Parámetros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene la información.

    columna : str
        Nombre de la variable temporal.

    Salida
    -------
    Muestra un resumen con:

    - Tipo de dato.
    - Número de fechas distintas.
    - Fecha mínima.
    - Fecha máxima.
    - Valores nulos.
    """

    fechas = pd.to_datetime(df[columna])

    resumen = pd.DataFrame({
        "Variable": [columna],
        "Tipo de dato": [fechas.dtype],
        "Fechas distintas": [fechas.dt.date.nunique()],
        "Fecha mínima": [fechas.min()],
        "Fecha máxima": [fechas.max()],
        "Valores nulos": [fechas.isna().sum()]
    })

    display(resumen)
```
